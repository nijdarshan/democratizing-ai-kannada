```python
def tf_idf(corpus):
    unique = {word.lower() for sent in corpus for word in sent.split()}  
    tf = []
    num_docs = len(corpus)
    for i, doc in enumerate(corpus):
        tf.append({})
        for word in unique:
            tf[i][word] = 0
            for sent_w in doc.split():
                if sent_w.lower() == word:
                    tf[i][word] += 1
    idf = {}
    for word in unique:
        idf[word] = 0
        for i, score in enumerate(tf):
            if score[word] > 0:
                idf[word] += 1
                
    for id in idf:
        idf[id] = num_docs/idf[id]

    
    return tf, idf
                    
```

```python
import math
from collections import defaultdict

def tf_idf(corpus):
    # Step 1: Calculate term frequencies (TF)
    tf = []
    num_docs = len(corpus)
    
    for doc in corpus:
        tf_doc = defaultdict(int)
        words = doc.split()
        for word in words:
            tf_doc[word.lower()] += 1
        tf.append(tf_doc)
    
    # Step 2: Calculate document frequencies (DF) for IDF calculation
    df = defaultdict(int)
    for doc_tf in tf:
        for word in doc_tf:
            df[word] += 1
    
    # Step 3: Calculate inverse document frequencies (IDF)
    idf = {}
    for word, count in df.items():
        idf[word] = num_docs / count
    
    # Step 4: Calculate TF-IDF scores
    tfidf = []
    for doc_tf in tf:
        tfidf_doc = {}
        for word, count in doc_tf.items():
            tfidf_doc[word] = count * idf[word]
        tfidf.append(tfidf_doc)
    
    return tfidf

# # Example corpus
# corpus = [
#     "this is a sample document",
#     "this document is another sample document",
#     "and this is yet another example",
#     "example document is here",
#     "here is a different document example"
# ]

# Calculate TF-IDF
tfidf_scores = tf_idf(corpus)

# Convert the TF-IDF dictionary to a vector
unique_words = sorted({word.lower() for sent in corpus for word in sent.split()})
vector = []
for doc in tfidf_scores:
    vector.append([doc.get(word, 0.0) for word in unique_words])

# Display TF-IDF vectors
for vec in vector:
    print(vec)
```

```python
def tf_idf(corpus):
    # Step 1: Calculate term frequencies (TF) with normalization
    tf = []
    num_docs = len(corpus)
    
    for doc in corpus:
        tf_doc = defaultdict(int)
        words = doc.lower().split()
        total_terms = len(words)
        for word in words:
            tf_doc[word] += 1
        # Normalize term frequencies by total terms in the document
        for word in tf_doc:
            tf_doc[word] /= total_terms
        tf.append(tf_doc)
    
    # Step 2: Calculate document frequencies (DF) for IDF calculation
    df = defaultdict(int)
    for doc_tf in tf:
        for word in doc_tf:
            df[word] += 1
    
    # Step 3: Calculate inverse document frequencies (IDF) using provided formula
    idf = {}
    for word, count in df.items():
        idf[word] = num_docs / count
    
    # Step 4: Calculate TF-IDF scores
    tfidf = []
    for doc_tf in tf:
        tfidf_doc = {}
        for word, count in doc_tf.items():
            tfidf_doc[word] = count * idf[word]
        tfidf.append(tfidf_doc)
    
    return tfidf

# # Example corpus
# corpus = [
#     "This is a sample document.",
#     "This document is another sample document.",
#     "And this is yet another example.",
#     "Example document is here.",
#     "Here is a different document example."
# ]

# Calculate TF-IDF
tfidf_scores = tf_idf(corpus)

# Convert the TF-IDF dictionary to a vector
unique_words = sorted({word.lower() for sent in corpus for word in sent.split()})
vector = []
for doc in tfidf_scores:
    vector.append([doc.get(word, 0.0) for word in unique_words])

# Display TF-IDF vectors
for vec in vector:
    print(vec)
```
