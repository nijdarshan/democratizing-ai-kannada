```python
import re
import random
import json

def remove_html_chars(text):
    """
    Removes HTML tags and entities from the given text.
    """
    cleaned_text = re.sub(r'<[^>]+>', '', text)  # Remove HTML tags
    cleaned_text = re.sub(r'&[^;]+;', '', cleaned_text)  # Remove HTML entities
    return cleaned_text

def load_corpus_and_pick_sentences_for_words(file_path, words, num_sentences):
    """
    Loads a corpus, finds sentences with an exact match of given words, cleans HTML characters,
    and randomly picks a specified number of these sentences for each word.
    """
    results = {}
    with open(file_path, 'r', encoding='utf-8') as file:
        corpus = file.read()
    
    sentences = re.split(r'[.\n]+', corpus)
    
    for word in words:
        word_boundary_regex = r'\b' + re.escape(word) + r'\b'
        matching_sentences = [remove_html_chars(sentence.strip()) for sentence in sentences if re.search(word_boundary_regex, sentence)]
        
        if len(matching_sentences) < num_sentences:
            selected_sentences = matching_sentences
        else:
            selected_sentences = random.sample(matching_sentences, num_sentences)
        
        results[word] = selected_sentences
    
    return json.dumps(results, ensure_ascii=False, indent=4)
```

```python
file_path = r"C:\Users\ADMIN\Korpora\kowikitext\kowikitext_20200920.train"
words = ["수면", "사전", "말", "이상", "사고"]
num_sentences_to_pick = 200  # Specify the number of sentences you want to pick for each word

json_results = load_corpus_and_pick_sentences_for_words(file_path, words, num_sentences_to_pick)

# Save the JSON results to a file
output_file_path = r"5_words_sentences.json"
with open(output_file_path, 'w', encoding='utf-8') as file:
	file.write(json_results)
```
