```python
import pandas as pd
import json
import re

def parse_classification_results(text):
    """Parse the classification results text into a list of dictionaries"""
    # Split the text into individual JSON blocks
    json_blocks = re.split(r'\n={50,}\n', text)
    
    all_classifications = []
    
    for block in json_blocks:
        if not block.strip():
            continue
            
        try:
            data = json.loads(block.strip())
            word = data['word']
            
            # Extract each sentence and its classification
            for classification in data['classifications']:
                all_classifications.append({
                    'word': word,
                    'sentence': classification['sentence'],
                    'label': classification['label']
                })
                
        except json.JSONDecodeError:
            print(f"Could not parse block: {block[:100]}...")
            continue
            
    return pd.DataFrame(all_classifications)

def create_final_dataset(classification_text, meanings_file):
    # Read the meanings file
    meanings_df = pd.read_csv(meanings_file)
    
    # Parse classification results into DataFrame
    results_df = parse_classification_results(classification_text)
    
    # Filter out irrelevant classifications (label = -1)
    results_df = results_df[results_df['label'] != -1]
    
    # For each row in results_df, get the corresponding meaning based on the label
    def get_meanings(row):
        word_meanings = meanings_df[meanings_df['word'] == row['word']].iloc[0]
        label = row['label']
        
        if label > 0:  # Valid classification
            kannada_col = f'kannada_meaning_{label}'
            english_col = f'english_meaning_{label}'
            
            return pd.Series({
                'kannada_meaning': word_meanings[kannada_col],
                'english_meaning': word_meanings[english_col]
            })
        
        return pd.Series({
            'kannada_meaning': None,
            'english_meaning': None
        })
    
    # Add meanings to results
    final_df = results_df.join(results_df.apply(get_meanings, axis=1))
    
    # Reorder columns
    final_df = final_df[['word', 'sentence', 'label', 'kannada_meaning', 'english_meaning']]
    
    return final_df

# Read the classification results text
with open('classification_results.txt', 'r', encoding='utf-8') as f:
    classification_text = f.read()

# Create the final dataset
final_dataset = create_final_dataset(classification_text, 'kanPolyMeanings.csv')

# Save to CSV
final_dataset.to_csv('polysemy_dataset_with_meanings.csv', index=False, encoding='utf-8')

# Print sample of the dataset
print("\nSample of the final dataset:")
print(final_dataset.head())
print("\nDataset shape:", final_dataset.shape)
```

```python
import pandas as pd
import json
import re

def parse_classification_results(text):
    """Parse the classification results text into a list of dictionaries"""
    # Split the text into individual JSON blocks
    json_blocks = re.split(r'\n={50,}\n', text)
    
    all_classifications = []
    
    for block in json_blocks:
        if not block.strip():
            continue
            
        try:
            data = json.loads(block.strip())
            word = data['word']
            
            # Extract each sentence and its classification
            for classification in data['classifications']:
                # Only include if label is not -1
                if classification['label'] != -1:
                    all_classifications.append({
                        'word': word,
                        'sentence': classification['sentence'],
                        'label': classification['label']
                    })
        except json.JSONDecodeError:
            continue
    
    # Convert to DataFrame
    df = pd.DataFrame(all_classifications)
    return df

# Read the classification results text
with open('classification_results.txt', 'r', encoding='utf-8') as f:
    classification_text = f.read()

# Create initial dataset
initial_dataset = parse_classification_results(classification_text)

# Save to CSV
initial_dataset.to_csv('classified_sentences_llm.csv', index=False, encoding='utf-8')

# Print sample of the dataset
print("\nSample of the initial dataset:")
print(initial_dataset.head())
print("\nDataset shape:", initial_dataset.shape)
```

```python
import pandas as pd

# Read both CSVs
classified_df = pd.read_csv('classified_sentences_llm.csv', encoding='utf-8')
meanings_df = pd.read_csv('kanPolyMeanings.csv', encoding='utf-8')

def get_meaning(row, meanings_df):
    """Get the corresponding Kannada and English meanings based on word and label"""
    word_meanings = meanings_df[meanings_df['word'] == row['word']].iloc[0]
    label = row['label']
    
    # Get corresponding meaning columns
    kannada_col = f'kannada_meaning_{label}'
    english_col = f'english_meaning_{label}'
    
    # Return meanings (handle missing values)
    kannada_meaning = word_meanings.get(kannada_col, '')
    english_meaning = word_meanings.get(english_col, '')
    
    return pd.Series({
        'kannada_meaning': str(kannada_meaning).strip(),
        'english_meaning': str(english_meaning).strip()
    })

# Add meanings to each row
meanings_series = classified_df.apply(lambda row: get_meaning(row, meanings_df), axis=1)
final_df = pd.concat([classified_df, meanings_series], axis=1)

# Save to CSV
final_df.to_csv('polysemy_dataset_with_meanings.csv', index=False, encoding='utf-8')

# Print sample
print("\nSample of final dataset:")
print(final_df.head())
print("\nDataset shape:", final_df.shape)
```

```python
import pandas as pd

# Read both CSVs
classified_df = pd.read_csv('classified_sentences_cleaned.csv', encoding='utf-8')
meanings_df = pd.read_csv('kanPolyMeanings.csv', encoding='utf-8')

def get_meaning(row, meanings_df):
    """Get the corresponding Kannada and English meanings based on word and label"""
    word_meanings = meanings_df[meanings_df['word'] == row['word']].iloc[0]
    label = row['label']
    
    # Map labels to meaning numbers
    label_to_meaning = {
        1: 1,  # label 0 -> meaning_1
        2: 2,  # label 1 -> meaning_2
        3: 3,  # label 2 -> meaning_3
        4: 4   # label 3 -> meaning_4
    }
    
    meaning_num = label_to_meaning.get(label)
    if meaning_num is None:
        return pd.Series({'kannada_meaning': '', 'english_meaning': ''})
    
    # Get corresponding meaning columns
    kannada_col = f'kannada_meaning_{meaning_num}'
    english_col = f'english_meaning_{meaning_num}'
    
    # Return meanings (handle missing values)
    kannada_meaning = word_meanings.get(kannada_col, '')
    english_meaning = word_meanings.get(english_col, '')
    
    return pd.Series({
        'kannada_meaning': str(kannada_meaning).strip(),
        'english_meaning': str(english_meaning).strip()
    })

# Filter out label 4 entries
classified_df = classified_df[classified_df['label'] != 4]

# Add meanings to each row
meanings_series = classified_df.apply(lambda row: get_meaning(row, meanings_df), axis=1)
final_df = pd.concat([classified_df, meanings_series], axis=1)

# Save to CSV
final_df.to_csv('polysemy_dataset_with_meanings_cleaned.csv', index=False, encoding='utf-8')

# Print sample
print("\nSample of final dataset:")
print(final_df.head())
print("\nDataset shape:", final_df.shape)
```
