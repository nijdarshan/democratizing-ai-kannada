```python
import json
import re

# Load JSON data from a file
def load_json(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        return json.load(file)

# Clean text by removing English words and unwanted characters
def clean_text(text):
    # Regex to match English words and characters
    english_regex = re.compile(r'[a-zA-Z]+|[.,;:"\'!?()-]')
    # Remove English words and specified punctuation
    cleaned_text = english_regex.sub('', text)
    return cleaned_text

# Process all sentences and clean them
def process_data(data):
    cleaned_data = {}
    for key, sentences in data.items():
        cleaned_sentences = [clean_text(sentence) for sentence in sentences]
        cleaned_data[key] = cleaned_sentences
    return cleaned_data

# Save data back to JSON
def save_json(data, output_file_path):
    with open(output_file_path, 'w', encoding='utf-8') as file:
        json.dump(data, file, ensure_ascii=False, indent=4)


input_file_path = 'data.json'
output_file_path = 'processed.json'

# Load data
data = load_json(input_file_path)

# Process and clean data
cleaned_data = process_data(data)

# Save cleaned data
save_json(cleaned_data, output_file_path)
print(f"Cleaned data has been saved to {output_file_path}")

```

```python
import json
import csv
import random

# Load JSON data from a file
def load_json(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        return json.load(file)

# Randomly pick 20 sentences from a list of sentences
def pick_random_sentences(sentences, num_sentences=50):
    if len(sentences) > num_sentences:
        return random.sample(sentences, num_sentences)
    return sentences

# Write data to a CSV file
def save_to_csv(data, output_file_path):
    with open(output_file_path, mode='w', newline='', encoding='utf-8') as file:
        writer = csv.writer(file)
        writer.writerow(['Key', 'Sentences'])  # Writing header
        for key, sentences in data.items():
            # Join the list of sentences into a single string, separating by a specific delimiter
            writer.writerow([key, '|'.join(sentences)])

# Main function to load, process, and save data
input_file_path = 'processed.json'
output_file_path = 'kannada_polysemy.csv'

# Load data
data = load_json(input_file_path)

# Process and randomly pick sentences
selected_sentences = {key: pick_random_sentences(sentences) for key, sentences in data.items()}

# Save selected sentences to CSV
save_to_csv(selected_sentences, output_file_path)
print(f"Data has been saved to {output_file_path}")

```

```python
import json

def read_json_and_write_sentences(input_json_path, output_txt_path):
    # Open and read the JSON file
    with open(input_json_path, 'r', encoding='utf-8') as json_file:
        data = json.load(json_file)

    # Open the text file where you want to write the sentences
    with open(output_txt_path, 'w', encoding='utf-8') as text_file:
        # Iterate over each key in the JSON data
        for sentences in data.values():  # Assuming each key's value is a list of sentences
            # Write each sentence to a new line in the text file
            for sentence in sentences:
                text_file.write(sentence + '\n')

# Define the paths to your input JSON and output text files
input_json_path = 'processed.json'
output_txt_path = 'sentences.txt'

# Call the function
read_json_and_write_sentences(input_json_path, output_txt_path)
```
