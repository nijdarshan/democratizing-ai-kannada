```python
import re
import random

def process_file(file_path, keyword):
    sentence_buffer = []
    all_sentences = []  # Store all sentences containing the keyword
    keyword_pattern = re.compile(r'\b' + re.escape(keyword) + r'\b', re.IGNORECASE)
    sentence_end_pattern = re.compile(r'[.!?]')

    def extract_sentences(buffer):
        content = ''.join(buffer)
        potential_sentences = re.split(sentence_end_pattern, content)
        return potential_sentences[:-1], potential_sentences[-1]

    with open(file_path, 'r', encoding='utf-8') as file:
        for line in file:
            sentence_buffer.append(line)
            if '.' in line or '?' in line or '!' in line:  # Likely end of sentence
                completed_sentences, carry_over = extract_sentences(sentence_buffer)
                sentence_buffer = [carry_over]  # Start new buffer with the carryover

                for sentence in completed_sentences:
                    if keyword_pattern.search(sentence):
                        all_sentences.append(sentence + '.')  # Re-add the end punctuation

    # Check any remaining content in the buffer after file processing
    if sentence_buffer:
        remaining_sentences, _ = extract_sentences(sentence_buffer)
        for sentence in remaining_sentences:
            if keyword_pattern.search(sentence):
                all_sentences.append(sentence + '.')

    # Randomly select 200 sentences from all found sentences, if there are at least 200
    if len(all_sentences) > 200:
        return random.sample(all_sentences, 200)
    else:
        return all_sentences

# Example usage
keyword = 'ಗುಂಟೆ'
file_path = 'D:/mscp/dai/data/kn (1).txt/kn_2.txt'
found_sentences = process_file(file_path, keyword)
print(found_sentences)
```

```python
import re
import random

# Example words dictionary with keywords initialized to empty lists
words = {'ಗುಂಟೆ': [], 'ಯಜಮಾನ': [], 'ಭಾವ': []}

def find_sentences(file_path, keywords):
    # Create a combined regex pattern for all keywords to find exact matches
    keyword_pattern = re.compile(r'\b(' + '|'.join(re.escape(k) for k in keywords) + r')\b', re.IGNORECASE)
    sentence_end_pattern = re.compile(r'[.!?]')
    
    # Initialize the sentence processing
    buffer = ""
    all_matches = {key: [] for key in keywords}

    # Process file line by line
    with open(file_path, 'r', encoding='utf-8') as file:
        for line in file:
            # Accumulate lines to buffer until we encounter a sentence end
            buffer += line
            if sentence_end_pattern.search(line):
                # Split buffer into sentences
                parts = sentence_end_pattern.split(buffer)
                # All but the last part are complete sentences
                for part in parts[:-1]:
                    sentence = part.strip() + '.'
                    # Check each sentence for keywords
                    matches = set(keyword_pattern.findall(sentence))
                    for match in matches:
                        all_matches[match].append(sentence)
                # The last part is the beginning of the next sentence
                buffer = parts[-1]
    
    # Final check for any remaining content in buffer
    if buffer.strip():
        sentence = buffer.strip() + '.'
        matches = set(keyword_pattern.findall(sentence))
        for match in matches:
            all_matches[match].append(sentence)

    # Populate the words dictionary with up to 200 random sentences or all found ones
    for keyword in keywords:
        found_sentences = all_matches[keyword]
        if len(found_sentences) > 200:
            words[keyword] = random.sample(found_sentences, 200)
        else:
            words[keyword] = found_sentences

    return words

# Example usage
file_path = 'D:/mscp/dai/data/kn (1).txt/kn_2.txt'
keywords = list(words.keys())
updated_words = find_sentences(file_path, keywords)
print(updated_words)

```

```python
import re
import random

# Example words dictionary with keywords initialized to empty lists
words = {'ವರ್ಷ': []}

def find_sentences(file_path, keywords):
    # Regex pattern to find exact matches, using negative lookarounds to ensure word boundaries
    patterns = {key: re.compile(r'(?<![\u0C80-\u0CFF])' + re.escape(key) + r'(?![\u0C80-\u0CFF])', re.IGNORECASE) for key in keywords}
    sentence_end_pattern = re.compile(r'[.!?]')

    all_matches = {key: [] for key in keywords}  # Dict to hold matches for each keyword
    
    with open(file_path, 'r', encoding='utf-8') as file:
        buffer = ""
        for line in file:
            buffer += line
            if sentence_end_pattern.search(line):
                parts = sentence_end_pattern.split(buffer)
                for part in parts[:-1]:
                    sentence = part.strip() + '.'
                    for key, pattern in patterns.items():
                        if pattern.search(sentence):
                            all_matches[key].append(sentence)
                buffer = parts[-1]
    
    if buffer.strip():
        sentence = buffer.strip() + '.'
        for key, pattern in patterns.items():
            if pattern.search(sentence):
                all_matches[key].append(sentence)

    # Populate the words dictionary with up to 200 random sentences or all found ones
    for keyword in keywords:
        found_sentences = all_matches[keyword]
        if len(found_sentences) > 200:
            words[keyword] = random.sample(found_sentences, 200)
        else:
            words[keyword] = found_sentences

    return words

# Example usage
file_path = 'D:/mscp/dai/data/kn (1).txt/kn_2.txt'
keywords = list(words.keys())
updated_words = find_sentences(file_path, keywords)
print(updated_words)
```
