```python
import os
```

```python
with open('samples.txt', 'r', encoding="utf8") as file:
    for line in file:
        word = line.split()
        if len(word) > 3:
            if word[3] == 'yes':
                with open('candidate_words.txt', "a") as cw_file:
                    cw_file.write(word[1]+"\n")
            elif word[3] == 'maybe':
                with open('maybe_candidate_words.txt', "a") as cw_file:
                    cw_file.write(word[1]+"\n")
    
```
