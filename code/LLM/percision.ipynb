```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import precision_recall_fscore_support, confusion_matrix

# Load your dataset
data = df

# Step 1: Calculate metrics per word per model
results = []

for model in data['model'].unique():
    model_data = data[data['model'] == model]
    for word in model_data['word'].unique():
        word_data = model_data[model_data['word'] == word]
        
        # True sense distribution (prior)
        prior_counts = word_data['true_sense'].value_counts(normalize=True)
        
        # Calculate precision and recall for true_sense vs predicted_sense
        precision_pred, recall_pred, _, _ = precision_recall_fscore_support(
            word_data['true_sense'], word_data['predicted_sense'], average=None, labels=word_data['true_sense'].unique())
        
        # Calculate precision and recall for true_sense vs llm_annotation
        precision_llm, recall_llm, _, _ = precision_recall_fscore_support(
            word_data['true_sense'], word_data['llm_annotation'], average=None, labels=word_data['true_sense'].unique())

        # Collect the results
        for i, sense in enumerate(word_data['true_sense'].unique()):
            results.append({
                "model": model,
                "word": word,
                "sense": sense,
                "prior": prior_counts.get(sense, 0),
                "precision_pred": precision_pred[i],
                "recall_pred": recall_pred[i],
                "precision_llm": precision_llm[i],
                "recall_llm": recall_llm[i]
            })

# Convert results to DataFrame for display
metrics_df = pd.DataFrame(results)

# Step 2: Plot agreement per model

for model in metrics_df['model'].unique():
    model_metrics = metrics_df[metrics_df['model'] == model]

    # Agreement bar chart: precision and recall for predicted vs llm annotations
    fig, ax = plt.subplots(1, 2, figsize=(14, 6))
    sns.barplot(data=model_metrics, x='sense', y='precision_pred', hue='word', ax=ax[0])
    sns.barplot(data=model_metrics, x='sense', y='recall_pred', hue='word', ax=ax[0])
    ax[0].set_title(f"{model} - Precision and Recall: True Sense vs Predicted Sense")
    ax[0].set_ylabel("Score")
    
    sns.barplot(data=model_metrics, x='sense', y='precision_llm', hue='word', ax=ax[1])
    sns.barplot(data=model_metrics, x='sense', y='recall_llm', hue='word', ax=ax[1])
    ax[1].set_title(f"{model} - Precision and Recall: True Sense vs LLM Annotation")
    ax[1].set_ylabel("Score")

    plt.tight_layout()
    plt.show()

# Step 3: Plot Confusion Matrix for Agreement per Model

for model in data['model'].unique():
    fig, axs = plt.subplots(1, 2, figsize=(12, 5))
    
    # Predicted Sense Agreement
    cm_pred = confusion_matrix(data[data['model'] == model]['true_sense'], data[data['model'] == model]['predicted_sense'])
    sns.heatmap(cm_pred, annot=True, fmt='d', cmap='Blues', ax=axs[0])
    axs[0].set_title(f"{model} - True Sense vs Predicted Sense")
    axs[0].set_xlabel("Predicted Sense")
    axs[0].set_ylabel("True Sense")
    
    # LLM Annotation Agreement
    cm_llm = confusion_matrix(data[data['model'] == model]['true_sense'], data[data['model'] == model]['llm_annotation'])
    sns.heatmap(cm_llm, annot=True, fmt='d', cmap='Greens', ax=axs[1])
    axs[1].set_title(f"{model} - True Sense vs LLM Annotation")
    axs[1].set_xlabel("LLM Annotation")
    axs[1].set_ylabel("True Sense")

    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.metrics import precision_recall_fscore_support

# Load your dataset
data = df

# Prepare results storage
results = []

# Loop through each model and word to calculate metrics
for model in data['model'].unique():
    model_data = data[data['model'] == model]
    for word in model_data['word'].unique():
        word_data = model_data[model_data['word'] == word]
        
        # Calculate prior (distribution of true_sense per class for this word)
        prior_counts = word_data['true_sense'].value_counts(normalize=True)
        
        # Calculate precision and recall for true_sense vs predicted_sense
        precision_pred, recall_pred, _, _ = precision_recall_fscore_support(
            word_data['true_sense'], word_data['predicted_sense'], average=None, labels=word_data['true_sense'].unique())
        
        # Calculate precision and recall for true_sense vs llm_annotation
        precision_llm, recall_llm, _, _ = precision_recall_fscore_support(
            word_data['true_sense'], word_data['llm_annotation'], average=None, labels=word_data['true_sense'].unique())

        # Collect the results per sense for the word and model
        for i, sense in enumerate(word_data['true_sense'].unique()):
            results.append({
                "model": model,
                "word": word,
                "sense": sense,
                "prior": prior_counts.get(sense, 0),  # Get the prior for this sense
                "precision_pred": precision_pred[i],
                "recall_pred": recall_pred[i],
                "precision_llm": precision_llm[i],
                "recall_llm": recall_llm[i]
            })

# Convert results to DataFrame for organized display
metrics_df = pd.DataFrame(results)

# Display the final table in a nice format
print("Per Word, Per Model Metrics with Prior:")
print(metrics_df)
```

```python
import pandas as pd
from sklearn.metrics import precision_recall_fscore_support

# Sample data for demonstration purposes
data = df

# Prepare results storage
results_pred = []
results_llm = []
combined_results = []

# Loop through each model and word to calculate metrics
for model in data['model'].unique():
    model_data = data[data['model'] == model]
    for word in model_data['word'].unique():
        word_data = model_data[model_data['word'] == word]
        
        # Calculate prior (distribution of true_sense per class for this word)
        prior_counts = word_data['true_sense'].value_counts(normalize=True)

        # Precision and recall for true_sense vs predicted_sense
        precision_pred, recall_pred, _, _ = precision_recall_fscore_support(
            word_data['true_sense'], word_data['predicted_sense'], average=None, labels=word_data['true_sense'].unique())
        
        # Precision and recall for true_sense vs llm_annotation
        precision_llm, recall_llm, _, _ = precision_recall_fscore_support(
            word_data['true_sense'], word_data['llm_annotation'], average=None, labels=word_data['true_sense'].unique())

        # Collect the results per sense for each table
        for i, sense in enumerate(word_data['true_sense'].unique()):
            # Table 1: true_sense vs predicted_sense
            results_pred.append({
                "Model": model,
                "Word": word,
                "Sense": sense,
                "Prior": round(prior_counts.get(sense, 0), 2),
                "Precision (predicted_sense)": round(precision_pred[i], 2),
                "Recall (predicted_sense)": round(recall_pred[i], 2),
            })
            # Table 2: true_sense vs llm_annotation
            results_llm.append({
                "Model": model,
                "Word": word,
                "Sense": sense,
                "Prior": round(prior_counts.get(sense, 0), 2),
                "Precision (llm_annotation)": round(precision_llm[i], 2),
                "Recall (llm_annotation)": round(recall_llm[i], 2),
            })
            # Table 3: Combined Summary
            combined_results.append({
                "Model": model,
                "Word": word,
                "Sense": sense,
                "Prior": round(prior_counts.get(sense, 0), 2),
                "Precision (predicted_sense)": round(precision_pred[i], 2),
                "Recall (predicted_sense)": round(recall_pred[i], 2),
                "Precision (llm_annotation)": round(precision_llm[i], 2),
                "Recall (llm_annotation)": round(recall_llm[i], 2),
            })

# Function to print DataFrame in Markdown format
def print_markdown_table(data, title):
    print(f"### {title}\n")
    print(data.to_markdown(index=False))
    print("\n")

# Convert results to DataFrames for display
table1_df = pd.DataFrame(results_pred)
table2_df = pd.DataFrame(results_llm)
table3_df = pd.DataFrame(combined_results)

# Print Markdown tables
print_markdown_table(table1_df, "Table 1: `true_sense` vs `predicted_sense`")
print_markdown_table(table2_df, "Table 2: `true_sense` vs `llm_annotation`")
print_markdown_table(table3_df, "Table 3: Combined Summary for `predicted_sense` and `llm_annotation`")
```

```python
import pandas as pd
from sklearn.metrics import precision_recall_fscore_support

# Sample data for demonstration purposes
data = df

# Prepare results storage
model_results = {}

# Loop through each model and word to calculate metrics
for model in data['model'].unique():
    model_data = data[data['model'] == model]
    model_results[model] = []
    
    for word in model_data['word'].unique():
        word_data = model_data[model_data['word'] == word]
        
        # Calculate prior (distribution of true_sense per class for this word)
        prior_counts = word_data['true_sense'].value_counts(normalize=True)

        # Precision and recall for true_sense vs predicted_sense
        precision_pred, recall_pred, _, _ = precision_recall_fscore_support(
            word_data['true_sense'], word_data['predicted_sense'], average=None, labels=word_data['true_sense'].unique())
        
        # Precision and recall for true_sense vs llm_annotation
        precision_llm, recall_llm, _, _ = precision_recall_fscore_support(
            word_data['true_sense'], word_data['llm_annotation'], average=None, labels=word_data['true_sense'].unique())

        # Collect the combined results per sense for this model and word
        for i, sense in enumerate(word_data['true_sense'].unique()):
            model_results[model].append({
                "Word": word,
                "Sense": sense,
                "Prior": round(prior_counts.get(sense, 0), 2),
                "Precision (predicted_sense)": round(precision_pred[i], 2),
                "Recall (predicted_sense)": round(recall_pred[i], 2),
                "Precision (llm_annotation)": round(precision_llm[i], 2),
                "Recall (llm_annotation)": round(recall_llm[i], 2),
            })

# Function to print DataFrame in Markdown format
def print_markdown_table(data, title):
    print(f"### {title}\n")
    print(data.to_markdown(index=False))
    print("\n")

# Generate and print tables for each model in Markdown format
for model, results in model_results.items():
    model_df = pd.DataFrame(results)
    print_markdown_table(model_df, f"Model: {model} - Combined Metrics for `predicted_sense` and `llm_annotation`")
```
