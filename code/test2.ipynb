```python
from polyglot.text import Text

# Sample Hindi text
text = "लड़के स्कूल जा रहे हैं।"

# Create a Polyglot Text object
polyglot_text = Text(text, hint_language_code='hi')

# Lemmatize the text
lemmatized_words = [word.lemma for word in polyglot_text.words]

# Print the original and lemmatized text
print("Original Text:", text)
print("Lemmatized Text:", " ".join(lemmatized_words))
```
