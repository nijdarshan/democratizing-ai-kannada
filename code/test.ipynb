```python
def assign_sense_numbers(group):
    # Create a mapping of unique meanings to sense numbers
    unique_meanings = group['english_meaning'].unique()
    meaning_to_sense = {meaning: idx+1 for idx, meaning in enumerate(unique_meanings)}
    
    # Map each meaning to its sense number
    group['sense'] = group['english_meaning'].map(meaning_to_sense)
    return group

# Group by word and apply the numbering
combined_v0 = combined_v0.groupby('word').apply(assign_sense_numbers).reset_index(drop=True)

# Display the first few rows to verify
print(combined_v0[['word', 'english_meaning', 'sense']].head(100))
```

```python
new_combined_v1 = combined_v0[["word", "sentence", "sense", "kannada_meaning", "english_meaning"]]
```

```python
new_combined_v1 = new_combined_v1.rename(columns={"word": "lemma"})
```

```python
# Group by lemma and get the maximum sense number for each
sense_counts = new_combined_v1.groupby('lemma')['sense'].max().sort_values(ascending=False)

# Display the results
print("Number of senses per lemma:")
print(sense_counts)

# Optional: Get some summary statistics
print("\nSummary:")
print(f"Total number of lemmas: {len(sense_counts)}")
print(f"Maximum number of senses for any lemma: {sense_counts.max()}")
print(f"Average number of senses per lemma: {sense_counts.mean():.2f}")
```

```python
# Filter to keep only lemmas with more than 1 sense
multi_sense_df = new_combined_v1[new_combined_v1['lemma'].isin(sense_counts[sense_counts > 1].index)]

# Count sentences per word per sense
sentence_counts = multi_sense_df.groupby(['lemma', 'sense']).size().reset_index(name='sentence_count')

# Sort by lemma and sense for better readability
sentence_counts = sentence_counts.sort_values(['lemma', 'sense'])

print("Number of sentences per lemma per sense:")
print(sentence_counts)

# Optional: Summary statistics
print("\nSummary:")
print(f"Total number of multi-sense lemmas: {len(sentence_counts['lemma'].unique())}")
print(f"Average sentences per sense: {sentence_counts['sentence_count'].mean():.2f}")
```

```python
def find_word_position(row):
    sentence = row['sentence']
    lemma = row['lemma']
    
    # Find the start position of the lemma in the sentence
    start = sentence.find(lemma)
    if start != -1:
        end = start + len(lemma)
    else:
        # If lemma not found exactly, set to -1
        start = -1
        end = -1
    
    return pd.Series({'start': start, 'end': end})

# Add start and end positions
position_df = filtered_df.copy()
position_df[['start', 'end']] = filtered_df.apply(find_word_position, axis=1)

# Group by lemma and sense
final_counts = position_df.groupby(['lemma', 'sense']).size().reset_index(name='count')

# Print a few examples to verify
print("\nFirst few rows with position information:")
print(position_df[['lemma', 'sense', 'sentence', 'start', 'end']].head(10))

# Print summary of words and their counts
print("\nCounts per lemma and sense:")
print(final_counts)
```

```python
def find_word_position(row):
    sentence = row['sentence']
    lemma = row['lemma']
    
    # Find the start position of the lemma in the sentence
    start = sentence.find(lemma)
    if start != -1:
        # For Kannada words, we need to count the actual characters
        # not the byte positions
        end = start + len(lemma)
        # Verify the extracted word matches the lemma
        extracted_word = sentence[start:end]
        if extracted_word != lemma:
            start = -1
            end = -1
    else:
        start = -1
        end = -1
    
    return pd.Series({'start': start, 'end': end})

# Add start and end positions
position_df = filtered_df.copy()
position_df[['start', 'end']] = filtered_df.apply(find_word_position, axis=1)

# Let's verify the extraction is correct
def verify_positions(row):
    if row['start'] != -1:
        extracted = row['sentence'][row['start']:row['end']]
        return extracted == row['lemma']
    return False

# Print some examples to verify
print("Sample results with extracted words:")
for _, row in position_df.head(10).iterrows():
    if row['start'] != -1:
        extracted = row['sentence'][row['start']:row['end']]
        print(f"\nLemma: {row['lemma']}")
        print(f"Sentence: {row['sentence']}")
        print(f"Start: {row['start']}, End: {row['end']}")
        print(f"Extracted word: {extracted}")
        print(f"Match correct: {row['lemma'] == extracted}")
```
